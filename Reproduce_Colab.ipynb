```python
from pathlib import Path
import zipfile
archive=Path('/content/Reviewer_Supplement.zip')
with zipfile.ZipFile(archive) as z:
    for name in z.namelist():
        assert (Path('/content/study')/name).resolve().is_relative_to(Path('/content/study'))
    z.extractall('/content/study')
```

```python
import subprocess,sys
subprocess.run([sys.executable,'-m','pip','install','-r','/content/study/requirements-analysis.txt'],check=True)
subprocess.run([sys.executable,'-m','venv','/content/embed-env'],check=True)
subprocess.run(['/content/embed-env/bin/python','-m','pip','install','-r','/content/study/requirements-embedding.txt'],check=True)
```

```python
subprocess.run([sys.executable,'/content/study/run_reproduction.py','--output','/content/reproduction','--embedding-python','/content/embed-env/bin/python'],check=True)
```
